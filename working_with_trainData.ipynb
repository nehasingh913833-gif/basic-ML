```python
#Feature Engineering

#Extract Year
train['current_date'] = pd.to_datetime(train['current_date'])
train['Year']=train['current_date'].dt.year
```

```python
#Extract Month
train['Month']=train['current_date'].dt.month
```

```python
#Month Name
train['Month_Name']=train['current_date'].dt.month_name()
```

```python
#Day
train['Day']=train['current_date'].dt.day
```

```python
#Day of Week
train['Weekday']=train['current_date'].dt.day_name()
```

```python
train['current_time'] = pd.to_datetime(train['current_time'], format='%I:%M:%S %p')
train['Hour']=train['current_time'].dt.hour
train['Minute']=train['current_time'].dt.minute
train['Second']=train['current_time'].dt.second
```

```python

#Peak Hour Feature
def peak(hour):

    if 6<=hour<=10:
        return "Morning Peak"

    elif 17<=hour<=21:
        return "Evening Peak"

    else:
        return "Non Peak"

train['Peak_Hour']=train['Hour'].apply(peak)
```

```python
#Time Zone Feature
def period(hour):

    if 0<=hour<6:
        return "Night"

    elif 6<=hour<12:
        return "Morning"

    elif 12<=hour<17:
        return "Afternoon"

    elif 17<=hour<21:
        return "Evening"

    else:
        return "Night"

train['Time_Period']=train['Hour'].apply(period)
```

```python
#Halt Difference
train['Halt_Difference']=train['mean_halt_times_destination']-train['mean_halt_times_source']
```

```python
#Coordinate Difference
train['Latitude_Diff']=abs(train['latitude_source']-train['latitude_destination'])

train['Longitude_Diff']=abs(train['longitude_source']-train['longitude_destination'])
```

```python
#Approximate Distance
train['Approx_Distance']=np.sqrt(
    (train['Latitude_Diff']**2)+
    (train['Longitude_Diff']**2)
)
```

<h1><b>Encoding + Feature Selection + Input/Output + Train-Test Split + Scaling

```python
# Remove Unnecessary Columns

# ==========================================
# Drop Unnecessary Columns
# ==========================================

drop_columns = [
    "id_code",                 # Unique ID
    "current_date",            # Already split into Year/Month/Day
    "current_time",            # Already split into Hour/Minute/Second
    "current_year",            # Same information as Year
    "Minute",                  # Usually less important than Hour
    "Second"                   # Usually adds noise
]

train.drop(columns=drop_columns, inplace=True)
```

```python
# Label Encoding

# Create separate LabelEncoder instances for each column

from sklearn.preprocessing import LabelEncoder

encoders = {}

categorical_columns = [
    "source_name",
    "destination_name",
    "train_name",
    "country_code_source",
    "country_code_destination",
    "current_day",
    "Month_Name",
    "Weekday",
    "Peak_Hour",
    "Time_Period",
    "target"
]

for col in categorical_columns:
    le = LabelEncoder()
    train[col] = le.fit_transform(train[col])
    encoders[col] = le
```

```python
#Feature Importance
from sklearn.ensemble import RandomForestClassifier

X = train.drop("target", axis=1)
y = train["target"]

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf.fit(X, y)

importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf.feature_importances_
}).sort_values(by="Importance", ascending=False)

importance
```

```python
#Plot Feature Importance
plt.figure(figsize=(10,8))

sns.barplot(
    data=importance,
    x="Importance",
    y="Feature"
)

plt.title("Feature Importance")

plt.show()
```

```python
#keep the top N features.

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

# Filter top_features to only include columns that currently exist in X
current_X_columns = X.columns.tolist()
top_features = [f for f in importance.head(18)["Feature"] if f in current_X_columns]

X = X[top_features]
```

```python
#Remove Low-Importance Features
low_features = importance[
    importance["Importance"] < 0.01
]["Feature"].tolist()

print(low_features)
```

```python
current_X_columns = X.columns.tolist()
features_to_drop = [f for f in low_features if f in current_X_columns]
X = X.drop(columns=features_to_drop)

print(X.shape)
```

```python
#Remove Highly Correlated Features
import numpy as np

corr_matrix = X.corr().abs()

upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

drop_corr = [
    column
    for column in upper.columns
    if any(upper[column] > 0.90)
]

print("Highly Correlated Columns:")
print(drop_corr)

X = X.drop(columns=drop_corr)

print(X.shape)
```

```python
#Recursive Feature Elimination (RFE)
from sklearn.feature_selection import RFE

selector = RFE(
    RandomForestClassifier(random_state=42),
    n_features_to_select=12
)

selector.fit(X, y)

selected_features = X.columns[selector.support_]

print(selected_features)
```

```python
#Keep Only the Best Features
X = X[selected_features]

print(X.head())
```

```python
#Train-Test Split
# ==========================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Shape :", X_train.shape)
print("Testing Shape  :", X_test.shape)

print("Training Target :", y_train.shape)
print("Testing Target  :", y_test.shape)
```

```python
#Feature Scaling

#Note: Scaling is required for KNN, SVM, and Logistic Regression. It is not required for tree-based models like Random Forest, XGBoost, or Decision Tree.

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print(X_train_scaled.shape)
print(X_test_scaled.shape)
```

```python
#Train Multiple Models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
#Create Evaluation Function
def evaluate_model(name, model, Xtr, Xte):

    model.fit(Xtr, y_train)

    y_pred = model.predict(Xte)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average="weighted")
    recall = recall_score(y_test, y_pred, average="weighted")
    f1 = f1_score(y_test, y_pred, average="weighted")

    print("="*60)
    print(name)
    print("="*60)

    print("Accuracy :", accuracy)
    print("Precision:", precision)
    print("Recall   :", recall)
    print("F1 Score :", f1)

    print("\nClassification Report\n")
    print(classification_report(y_test, y_pred))

    return [name, accuracy, precision, recall, f1]
```

```python
#Check Shape
print("Training Data :",X_train.shape)

print("Testing Data :",X_test.shape)
```

```python
#Train All Models
results = []

# Logistic Regression
results.append(
    evaluate_model(
        "Logistic Regression",
        LogisticRegression(max_iter=500),
        X_train_scaled,
        X_test_scaled
    )
)

# Decision Tree
results.append(
    evaluate_model(
        "Decision Tree",
        DecisionTreeClassifier(random_state=42),
        X_train,
        X_test
    )
)

# Random Forest
results.append(
    evaluate_model(
        "Random Forest",
        RandomForestClassifier(
            n_estimators=300,
            random_state=42
        ),
        X_train,
        X_test
    )
)

# Gradient Boosting
results.append(
    evaluate_model(
        "Gradient Boosting",
        GradientBoostingClassifier(random_state=42),
        X_train,
        X_test
    )
)

# AdaBoost
results.append(
    evaluate_model(
        "AdaBoost",
        AdaBoostClassifier(random_state=42),
        X_train,
        X_test
    )
)

# KNN
results.append(
    evaluate_model(
        "KNN",
        KNeighborsClassifier(n_neighbors=5),
        X_train_scaled,
        X_test_scaled
    )
)

# SVM
results.append(
    evaluate_model(
        "SVM",
        SVC(),
        X_train_scaled,
        X_test_scaled
    )
)

# Naive Bayes
results.append(
    evaluate_model(
        "Naive Bayes",
        GaussianNB(),
        X_train_scaled,
        X_test_scaled
    )
)
```

```python
#Compare Results
results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ]
)

results_df.sort_values(
    by="Accuracy",
    ascending=False,
    inplace=True
)

results_df
```

```python
#Plot Model Accuracy
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))

sns.barplot(
    data=results_df,
    x="Accuracy",
    y="Model"
)

plt.title("Model Accuracy Comparison")

plt.show()
```

```python
#Hyperparameter Tuning (Random Forest)
from sklearn.model_selection import GridSearchCV

params = {
    "n_estimators": [200, 300, 500],
    "max_depth": [10, 20, None],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

grid = GridSearchCV(
    RandomForestClassifier(random_state=42),
    params,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

grid.fit(X_train, y_train)

print("Best Parameters:")
print(grid.best_params_)

print("\nBest CV Accuracy:")
print(grid.best_score_)
```

```python
#Train Optimized Random Forest
best_rf = grid.best_estimator_

best_rf.fit(X_train, y_train)

pred = best_rf.predict(X_test)

print("Final Accuracy :", accuracy_score(y_test, pred))

print(classification_report(y_test, pred))
```

```python
#Feature Importance
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": best_rf.feature_importances_
})

importance = importance.sort_values(
    by="Importance",
    ascending=False
)

importance
```

```python
#Plot Feature Importance
plt.figure(figsize=(10,8))

sns.barplot(
    data=importance,
    x="Importance",
    y="Feature"
)

plt.title("Top Feature Importance")

plt.show()
```

```python
#Cross Validation
from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    best_rf,
    X,
    y,
    cv=10,
    scoring="accuracy"
)

print("Cross Validation Scores:")
print(scores)

print("\nAverage Accuracy:", scores.mean())
```
